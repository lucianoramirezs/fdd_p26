```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Datos sinteticos del modulo

```python
contract_levels = ['Wiki', 'README', 'Ejemplos', 'OpenAPI']
contract_strength = [2, 4, 6, 10]

change_labels = ['Agregar opcional', 'Endpoint nuevo', 'Quitar campo', 'Cambiar tipo']
change_risk = [1, 2, 9, 10]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].bar(contract_levels, contract_strength)
axes[0].set_title('Fuerza del contrato')
axes[1].barh(change_labels, change_risk)
axes[1].set_title('Riesgo de ruptura')
plt.tight_layout()
plt.show()
```

## 3. Lectura rapida

La idea de estas graficas no es una precision empirica perfecta, sino sintetizar decisiones del modulo:

- contratos fuertes reducen ambiguedad
- cambios breaking concentran riesgo
- releases necesitan metricas para ser confiables
