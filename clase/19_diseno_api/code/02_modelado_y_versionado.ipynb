## 1. Shapes fragiles vs durables

```python
fragil = {
    'reply': 'Hola',
    'tokens': 22
}

durable = {
    'id': 'msg_001',
    'output_text': 'Hola',
    'usage': {'input_tokens': 10, 'output_tokens': 12},
    'meta': {'model': 'chat-mini', 'trace_id': 'tr_123'}
}

fragil, durable
```

## 2. Cambios aditivos vs breaking

```python
cambios = [
    ('agregar campo opcional usage', 'non-breaking'),
    ('renombrar output_text a text', 'breaking'),
    ('quitar id', 'breaking'),
    ('agregar endpoint GET /v1/usage', 'non-breaking'),
    ('hacer obligatorio temperature', 'breaking')
]

for cambio, tipo in cambios:
    print(f'- {cambio}: {tipo}')
```

## 3. Paginacion simple

```python
jobs = [f'job_{i:03d}' for i in range(1, 16)]

def paginar_cursor(items, cursor=None, limit=5):
    start = 0 if cursor is None else int(cursor)
    chunk = items[start:start + limit]
    next_cursor = None if start + limit >= len(items) else str(start + limit)
    return {'data': chunk, 'next_cursor': next_cursor}

pagina_1 = paginar_cursor(jobs)
pagina_2 = paginar_cursor(jobs, cursor=pagina_1['next_cursor'])
pagina_1, pagina_2
```

## 4. Checklist de versionado

```python
checklist = [
    'El consumer viejo sigue parseando?',
    'El cambio es aditivo?',
    'Hay tests de contrato?',
    'Se puede deprecar antes de borrar?',
    'Hay metricas para detectar clientes rotos?'
]

for item in checklist:
    print('[ ]', item)
```
