```python
from jsonschema import validate, ValidationError
```

## 2. Schema esperado por el consumer

```python
response_schema = {
    'type': 'object',
    'required': ['id', 'output_text'],
    'properties': {
        'id': {'type': 'string'},
        'output_text': {'type': 'string'},
        'usage': {
            'type': 'object',
            'properties': {
                'input_tokens': {'type': 'integer'},
                'output_tokens': {'type': 'integer'}
            }
        }
    }
}
```

## 3. Producer sano vs producer con drift

```python
respuesta_sana = {
    'id': 'msg_001',
    'output_text': 'Hola',
    'usage': {'input_tokens': 10, 'output_tokens': 12}
}

respuesta_con_drift = {
    'id': 'msg_001',
    'text': 'Hola'
}

validate(instance=respuesta_sana, schema=response_schema)
print('Respuesta sana: OK')

try:
    validate(instance=respuesta_con_drift, schema=response_schema)
except ValidationError as e:
    print('Drift detectado:')
    print('-', e.message)
```
